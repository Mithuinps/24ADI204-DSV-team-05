WEEK 02 – KNOW YOUR DATA

Project: Moneyball – IPL Player Valuation Team Number: 05

Objective: To understand the structure, data types, missing values and quality of the IPL dataset.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_csv("IPL_Moneyball_Merged_2008_2026.csv")

print("Dataset loaded successfully!")
```

```python
print("FIRST 5 ROWS")
display(df.head())

print("LAST 5 ROWS")
display(df.tail())

print("RANDOM 5 ROWS")
display(df.sample(5))
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
print("Dataset Shape:", df.shape)
```

```python
print("DATASET COLUMN NAMES")

for i, col in enumerate(df.columns, 1):
    print(i, "-", col)
```

```python
print("DATASET INFORMATION")

df.info()
```

```python
print("DATA TYPES OF ALL COLUMNS")

display(df.dtypes)
```

```python
print("INTEGER COLUMNS:", df.select_dtypes(include='int').columns.tolist())

print("FLOAT COLUMNS:", df.select_dtypes(include='float').columns.tolist())

print("OBJECT COLUMNS:", df.select_dtypes(include='object').columns.tolist())
```

```python
print("MISSING VALUES IN EACH COLUMN")

display(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

display(missing_df)
```

```python
plt.figure(figsize=(14, 6))

sns.heatmap(df.isnull(), cbar=False, cmap='viridis')

plt.title("Missing Values in IPL Dataset")
plt.xlabel("Columns")
plt.ylabel("Records")

plt.show()
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)

if duplicate_count == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records are present.")
```

```python
print("UNIQUE VALUES IN EACH COLUMN")

for col in df.columns:
    print(col, ":", df[col].nunique())
```

```python
print("NUMERICAL DATA SUMMARY")

display(df.describe())
```

```python
print("CATEGORICAL DATA SUMMARY")

display(df.describe(include='object'))
```

```python
print("PLAYER RECORDS BY SEASON")

season_count = df['season'].value_counts().sort_index()

display(season_count)
```

```python
print("TOP 10 PLAYERS BY NUMBER OF RECORDS")

player_count = df['player'].value_counts().head(10)

display(player_count)
```

```python
plt.figure(figsize=(12, 5))

season_count.plot(kind='bar')

plt.title("IPL Player Records Across Seasons")
plt.xlabel("Season")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
summary = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Total Columns',
        'Total Missing Values',
        'Duplicate Records',
        'Unique Players',
        'Unique Seasons'
    ],
    'Value': [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df['player'].nunique(),
        df['season'].nunique()
    ]
})

display(summary)
```

```python
print("WEEK 02 - KNOW YOUR DATA")
print("--------------------------------")

print("Dataset Shape:", df.shape)

print("Total Records:", len(df))

print("Total Attributes:", len(df.columns))

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Records:", df.duplicated().sum())

print("Unique Players:", df['player'].nunique())

print("Unique Seasons:", df['season'].nunique())

print("--------------------------------")
print("WEEK 02 DATA INSPECTION COMPLETED!")
```
